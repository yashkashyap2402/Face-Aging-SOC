# Week 7 & 8 - Face Aging Comparison

This notebook compares the original face image with the AI-generated aged face. The objective is to visually evaluate how well the age progression model preserves facial identity while introducing realistic aging effects.

```python
import matplotlib.pyplot as plt
from PIL import Image
from google.colab import files
```

## Step 1: Upload the Original Face Image

```python
print("Upload the original image")
original_upload = files.upload()
```

```python
original_name = list(original_upload.keys())[0]
original = Image.open(original_name)

plt.imshow(original)
plt.axis("off")
plt.title("Original Image")
plt.show()
```

## Step 2: Upload the Generated Aged Face

```python
print("Upload the aged face image")
aged_upload = files.upload()
```

```python
aged_name = list(aged_upload.keys())[0]
aged = Image.open(aged_name)

plt.imshow(aged)
plt.axis("off")
plt.title("Aged Face")
plt.show()
```

## Step 3: Display Side-by-Side Comparison

```python
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(original)
plt.title("Original")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(aged)
plt.title("Aged")
plt.axis("off")

plt.show()
```

## Step 4: Image Information

```python
print("Original Image Size:", original.size)
print("Aged Image Size:", aged.size)
```

## Step 5: Observations

The generated aged face should preserve the person's identity while introducing realistic aging characteristics such as:
