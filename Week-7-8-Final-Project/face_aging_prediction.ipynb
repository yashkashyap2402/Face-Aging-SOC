# Week 7 & 8 - Face Aging Prediction

This notebook demonstrates the workflow of generating an aged version of a face image using AI-based face aging tools. The process includes uploading an image, displaying it, generating an aged version, and saving the result for comparison.

```python
import matplotlib.pyplot as plt
from PIL import Image
from google.colab import files
```

## Step 1: Upload a Face Image

Upload a clear front-facing image of a person's face.

```python
uploaded = files.upload()
```

```python
image_name = list(uploaded.keys())[0]

original = Image.open(image_name)

plt.figure(figsize=(5,5))
plt.imshow(original)
plt.axis("off")
plt.title("Original Image")
plt.show()
```

## Step 2: Generate an Aged Face

Use an AI face-aging tool or model to generate an aged version of the uploaded image.

Example prompt:

"Generate a realistic aged version of this person's face, approximately 30 years older, while preserving facial identity, expression, pose, hairstyle, and lighting."

Save the generated image as:

aged_face.png

## Step 3: Upload the Generated Image

After generating the aged face using the AI tool, upload the generated image.

```python
aged_uploaded = files.upload()
```

```python
aged_name = list(aged_uploaded.keys())[0]

aged = Image.open(aged_name)

plt.figure(figsize=(5,5))
plt.imshow(aged)
plt.axis("off")
plt.title("Generated Aged Face")
plt.show()
```

## Step 4: Save the Generated Image

```python
aged.save("aged_face.png")

print("Generated image saved successfully.")
```

## Summary

In this notebook, we:

- Uploaded an original face image.
- Used an AI face-aging tool to generate an aged version.
- Uploaded the generated image.
- Saved the final aged image for comparison.
