# Week 6 - Age Conditional GAN

This notebook introduces Age Conditional Generative Adversarial Networks (Age-cGAN) and explains how conditional information such as age is used to generate facial images of different ages.

## What is an Age Conditional GAN?

An Age Conditional GAN is a Generative Adversarial Network that generates images based on a specified age condition. Instead of generating random faces, the generator receives both a random latent vector and a target age label to produce a face corresponding to that age.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Components of Age Conditional GAN

An Age Conditional GAN consists of:

1. Generator
2. Discriminator
3. Latent Vector
4. Age Label (Condition)

```python
components = [
    "Generator",
    "Discriminator",
    "Latent Vector",
    "Age Label"
]

for component in components:
    print(component)
```

## Latent Vector

The latent vector is a random numerical representation used as the starting point for image generation.

```python
latent_vector = np.random.randn(100)

print(latent_vector.shape)
```

## Age Label

The age label is provided to control the age of the generated face.

```python
age = 60

print("Target Age:", age)
```

## Combining Latent Vector and Age

The latent vector and age label together form the input to the Generator.

```python
combined_input = np.append(latent_vector, age)

print("Combined Input Shape:", combined_input.shape)
```

## Generator

The Generator creates a synthetic face using:

- Latent Vector
- Age Label

```python
generated_face = np.random.rand(64,64)

plt.imshow(generated_face,cmap="gray")
plt.title("Generated Face")
plt.axis("off")
plt.show()
```

## Discriminator

The Discriminator receives:

- Image
- Age Label

It predicts whether the image is real or generated.

```python
images = [
    "Real Face",
    "Generated Face"
]

for image in images:
    print(image)
```

## Training Process

1. Generate random latent vector.
2. Provide target age.
3. Generator creates a face.
4. Discriminator checks whether it is real.
5. Generator improves using discriminator feedback.
6. Repeat until realistic images are produced.

```python
steps = [
    "Random Noise",
    "Age Condition",
    "Generator",
    "Generated Face",
    "Discriminator",
    "Loss Calculation",
    "Update Networks"
]

for step in steps:
    print(step)
```

## Difference Between GAN and Age Conditional GAN

GAN:
- Generates random images.
- No control over output.

Age Conditional GAN:
- Generates images for a specified age.
- Produces controllable face aging results.

```python
comparison = {
    "GAN": "Random Image Generation",
    "Age Conditional GAN": "Age Controlled Image Generation"
}

for key, value in comparison.items():
    print(f"{key} : {value}")
```
