This notebook introduces the fundamentals of Generative Adversarial Networks (GANs), their architecture, components, and training process for image generation.

## What is a GAN?

A Generative Adversarial Network (GAN) is a deep learning model introduced by Ian Goodfellow in 2014. It consists of two neural networks that compete with each other:

- Generator (G)
- Discriminator (D)

The Generator creates fake images, while the Discriminator tries to distinguish between real and fake images. Through this competition, the Generator gradually learns to generate realistic images.

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Components of GAN

A GAN consists of two main networks:

1. Generator
2. Discriminator

```python
components = [
    "Generator",
    "Discriminator"
]

for component in components:
    print(component)
```

## Generator

The Generator takes a random noise vector as input and generates a synthetic image. Its objective is to create images that are realistic enough to fool the Discriminator.

```python
noise = np.random.randn(100)

print("Noise Vector Shape:", noise.shape)
```

## Fake Image Generation (Illustration)

In a real GAN, the Generator converts the noise vector into an image. Here, we use a random matrix only to illustrate the concept.

```python
fake_image = np.random.rand(28, 28)

plt.imshow(fake_image, cmap="gray")
plt.title("Generated Image (Illustration)")
plt.axis("off")
plt.show()
```

## Discriminator

The Discriminator receives both real images and generated images. Its task is to predict whether an input image is real or fake.

```python
images = [
    "Real Image",
    "Fake Image"
]

for image in images:
    print(image)
```

## Adversarial Training

During training:

1. The Generator creates fake images.
2. The Discriminator classifies images as real or fake.
3. The Generator learns from the Discriminator's feedback.
4. Both networks improve together through repeated training.

```python
training_steps = [
    "Generate Noise",
    "Generator",
    "Fake Image",
    "Discriminator",
    "Calculate Loss",
    "Update Generator",
    "Update Discriminator"
]

for step in training_steps:
    print(step)
```

## GAN Loss

The Generator tries to fool the Discriminator, while the Discriminator tries to correctly classify real and fake images.

Both networks are trained alternately until the Generator produces realistic images.

## Advantages of GANs

- Generates realistic images
- Learns complex data distributions
- Useful for data augmentation
- Can generate high-quality synthetic images

## Limitations of GANs

- Difficult to train
- Training instability
- Mode collapse
- Sensitive to hyperparameters
