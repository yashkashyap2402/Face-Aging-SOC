# Face Datasets

```python
from google.colab import files
from PIL import Image
import matplotlib.pyplot as plt
```

## UTKFace Dataset

```python
uploaded = files.upload()
```

```python
image_name = list(uploaded.keys())[0]

img = Image.open(image_name)

plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
print("Image Size:", img.size)
print("Image Mode:", img.mode)
```

#UTKFace is a facial image dataset containing age, gender, and race labels. It is commonly used for age estimation and face aging tasks.

#CelebA is a large-scale celebrity face dataset containing over 200,000 images with 40 facial attributes. It is widely used for face recognition, attribute prediction, and generative AI tasks.

## Conclusion: Both UTKFace and CelebA are important datasets in computer vision. UTKFace is mainly used for age-related tasks, while CelebA is used for facial attribute learning.
