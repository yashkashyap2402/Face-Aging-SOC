```python
from PIL import Image
import matplotlib.pyplot as plt
from google.colab import files
```

```python
uploaded = files.upload()
```

```python
img = Image.open("Screenshot 2026-06-23 143728.png")

img
```

```python
print("Format :", img.format)
print("Size :", img.size)
print("Mode :", img.mode)
```

```python
plt.imshow(img)
plt.axis("off")
plt.show()
```

```python
resized = img.resize((300,300))

plt.imshow(resized)
plt.axis("off")
plt.show()
```

```python
rotated = img.rotate(45)

plt.imshow(rotated)
plt.axis("off")
plt.show()
```

```python
cropped = img.crop((100,100,400,400))

plt.imshow(cropped)
plt.axis("off")
plt.show()
```

```python
flipped = img.transpose(Image.FLIP_LEFT_RIGHT)

plt.imshow(flipped)
plt.axis("off")
plt.show()
```

```python
gray = img.convert("L")

plt.imshow(gray,cmap="gray")
plt.axis("off")
plt.show()
```

```python
resized = resized.convert("RGB")
rotated = rotated.convert("RGB")

resized.save("resized_image.jpg")
rotated.save("rotated_image.jpg")
gray.save("gray_image.jpg")
```

```python
resized.save("resized_image.png")
rotated.save("rotated_image.png")
gray.save("gray_image.png")
```

```python
print(img.mode)
print(resized.mode)
print(rotated.mode)
print(gray.mode)
```

```python
thumbnail = img.copy()

thumbnail.thumbnail((150,150))

plt.imshow(thumbnail)
plt.axis("off")
plt.show()
```

```python
img = img.convert("RGB")

r, g, b = img.split()
```

```python
r,g,b = img.split()

plt.figure(figsize=(12,4))

plt.subplot(1,3,1)
plt.imshow(r,cmap="Reds")
plt.title("Red")

plt.subplot(1,3,2)
plt.imshow(g,cmap="Greens")
plt.title("Green")

plt.subplot(1,3,3)
plt.imshow(b,cmap="Blues")
plt.title("Blue")

plt.show()
```
