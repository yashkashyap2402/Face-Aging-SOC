```python
import matplotlib.pyplot as plt
from PIL import Image
from torchvision import transforms
from google.colab import files
```

```python
img = Image.open("Screenshot 2026-06-23 143728.png")
img = img.convert("RGB")

plt.imshow(img)
plt.axis("off")
plt.title("Original Image")
plt.show()
```

```python
resize = transforms.Resize((256,256))

resized = resize(img)

plt.imshow(resized)
plt.axis("off")
plt.show()
```

```python
center_crop = transforms.CenterCrop(200)

cropped = center_crop(img)

plt.imshow(cropped)
plt.axis("off")
plt.show()
```

```python
random_crop = transforms.RandomCrop(200)

cropped = random_crop(img)

plt.imshow(cropped)
plt.axis("off")
plt.show()
```

```python
flip = transforms.RandomHorizontalFlip(p=1)

flipped = flip(img)

plt.imshow(flipped)
plt.axis("off")
plt.show()
```

```python
vertical = transforms.RandomVerticalFlip(p=1)

vimg = vertical(img)

plt.imshow(vimg)
plt.axis("off")
plt.show()
```

```python
jitter = transforms.ColorJitter(
    brightness=0.5,
    contrast=0.5,
    saturation=0.5
)

jittered = jitter(img)

plt.imshow(jittered)
plt.axis("off")
plt.show()
```

```python
tensor = transforms.ToTensor()(img)

print(tensor.shape)
print(type(tensor))
```

```python
transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.5,0.5,0.5],
        std=[0.5,0.5,0.5]
    )
])

normalized = transform(img)

print(normalized.shape)
```

```python
transform = transforms.Compose([
    transforms.Resize((256,256)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.5,0.5,0.5],
        [0.5,0.5,0.5]
    )
])

output = transform(img)

print(output.shape)
```
