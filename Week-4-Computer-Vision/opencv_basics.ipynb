```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
from google.colab import files
```

```python
uploaded = files.upload()
```

```python
img = cv2.imread("Screenshot 2026-06-23 143728.png")
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

plt.imshow(img_rgb)
plt.axis("off")
plt.show()
```

```python
print(img.shape)
```

```python
resized = cv2.resize(img_rgb, (300,300))

plt.imshow(resized)
plt.axis("off")
plt.show()
```

```python
cropped = img_rgb[100:400,100:400]

plt.imshow(cropped)
plt.axis("off")
plt.show()
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

plt.imshow(gray,cmap="gray")
plt.axis("off")
plt.show()
```

```python
rectangle = img_rgb.copy()

cv2.rectangle(rectangle,
              (50,50),
              (250,250),
              (255,0,0),
              3)

plt.imshow(rectangle)
plt.axis("off")
plt.show()
```

```python
circle = img_rgb.copy()

cv2.circle(circle,
           (200,200),
           80,
           (0,255,0),
           4)

plt.imshow(circle)
plt.axis("off")
plt.show()
```

```python
line = img_rgb.copy()

cv2.line(line,
         (0,0),
         (400,400),
         (0,0,255),
         5)

plt.imshow(line)
plt.axis("off")
plt.show()
```

```python
text = img_rgb.copy()

cv2.putText(text,
            "Face Aging SOC",
            (40,40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (255,255,0),
            2)

plt.imshow(text)
plt.axis("off")
plt.show()
```

```python
flip = cv2.flip(img_rgb,1)

plt.imshow(flip)
plt.axis("off")
plt.show()
```

```python
blur = cv2.GaussianBlur(img_rgb,(11,11),0)

plt.imshow(blur)
plt.axis("off")
plt.show()
```

```python
edges = cv2.Canny(gray,100,200)

plt.imshow(edges,cmap="gray")
plt.axis("off")
plt.show()
```

```python
cv2.imwrite("processed_image.jpg",
            cv2.cvtColor(text,cv2.COLOR_RGB2BGR))
```
